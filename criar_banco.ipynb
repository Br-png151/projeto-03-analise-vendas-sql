Aprendizado novo:
- COUNT(*): conta os registros.
- AVG(): calcula uma média simples.
- Aqui usamos SUM() / COUNT() para calcular o ticket médio ponderado pelo faturamento de cada registro.

```python
conexao = sqlite3.connect("vendas.db")

consulta = pd.read_sql_query(
    """
    SELECT
        vendedor,
         SUM(quatidade*valor_unitario) AS faturamento_total,
         COUNT(*) AS quantidade_vendas,
         ROUND(
            AVG(quatidade*valor_unitario)
            ,2
        ) AS ticket_medio
    FROM vendas 
    GROUP BY vendedor
    ORDER BY ticket_medio DESC
    """, 
    conexao
)
conexao.close()
display(consulta)
```

```python
conexao = sqlite3.connect("vendas.db")

consulta = pd.read_sql_query(
    """
    SELECT
        vendedor,
         SUM(quatidade*valor_unitario) AS faturamento_total,
         CASE
            WHEN SUM(quatidade*valor_unitario)>= 5000
                THEN 'Alto'
            WHEN SUM(quatidade*valor_unitario)>=1000
                THEN 'Medio'
            ELSE 'Baixo'
        END AS classificaçao
    FROM vendas 
    GROUP BY vendedor
    ORDER BY faturamento_total DESC
    """, 
    conexao
)
conexao.close()
display(consulta)
```

O que estamos aprendendo?
- CASE: inicia uma condição.
- WHEN: verifica uma condição.
- THEN: define o resultado quando ela é verdadeira.
- ELSE: define o resultado quando nenhuma condição anterior é atendida.
- END: encerra a classificação.

```python
conexao = sqlite3.connect("vendas.db")

consulta = pd.read_sql_query(
    """
    SELECT
        produto,
         SUM(quatidade*valor_unitario) AS faturamento_total,
         CASE
            WHEN SUM(quatidade*valor_unitario)< 1000
                THEN 'Baixo'
            WHEN SUM(quatidade*valor_unitario)< 3000
                THEN 'Medio'
            ELSE 'ALTO'
        END AS classificaçao
    FROM vendas 
    GROUP BY produto
    ORDER BY faturamento_total DESC
    """, 
    conexao
)
conexao.close()
display(consulta)
```

Próxima etapa — LIKE
Agora vamos aprender a pesquisar textos parcialmente.

Imagine que você não sabe o nome completo do produto, mas sabe que começa com M.
O símbolo % representa qualquer sequência de caracteres.

Exemplos:
- 'M%' → começa com M.
- '%e' → termina com e.
- '%ou%' → contém ou.

```python
conexao = sqlite3.connect("vendas.db")
consulta = pd.read_sql_query (
    """
    SELECT * 
    FROM vendas
    WHERE produto LIKE 'M%'
    AND categoria = 'Eletrônicos';
    """,
    conexao
)
conexao.close()
display(consulta)
```

```python
conexao = sqlite3.connect("vendas.db")

consulta = pd.read_sql_query(
    """
    SELECT
        vendas.produto,
        vendas.categoria,
        vendas.valor_unitario,
        clientes.nome_cliente,
        clientes.cidade
    FROM vendas
    INNER JOIN clientes
        ON vendas.id_cliente = clientes.id_cliente;
    """,
    conexao
)

conexao.close()

display(consulta)
```

INNER JOIN: combina as duas tabelas.
ON: define como elas serão relacionadas.
id_cliente: é a chave que conecta os registros.

```python
conexao = sqlite3.connect("vendas.db")

consulta = pd.read_sql_query(
    """
    SELECT
        clientes.cidade,
         SUM(quatidade* valor_unitario) AS faturamento
    FROM vendas
    INNER JOIN clientes 
         ON vendas.id_cliente = clientes.id_cliente
    GROUP BY clientes.cidade
    ORDER BY faturamento DESC
    """, 
    conexao
)
conexao.close()
display(consulta)
```
